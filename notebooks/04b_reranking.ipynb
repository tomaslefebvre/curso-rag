Este notebook parte del concepto clave: el reranker **no busca**. Primero recuperamos candidatos y recién después el cross-encoder juzga cada par pregunta-documento.
Para mantenerlo simple, ejecutá primero el notebook del módulo 4 o copiá aquí su preparación.


```python
from sentence_transformers import CrossEncoder
reranker = CrossEncoder("cross-encoder/mmarco-mMiniLMv2-L12-H384-v1")

pregunta = "¿Qué significa el error VPN-1047?"
idxs = buscar_hibrido(pregunta, k=20)
pares = [[pregunta, chunks[i]] for i in idxs]
scores = reranker.predict(pares)
orden = np.argsort(-scores)

for pos in orden[:6]:
    idx = idxs[pos]
    print(round(float(scores[pos]),3), chunks[idx][:120].replace("\n"," "))
```
