## Experimento 1 — modelo inadecuado para español
Queremos observar un fallo silencioso.


```python
from sentence_transformers import SentenceTransformer, util

frases = [
    "El usuario no puede conectarse a la VPN",
    "Hay un problema con el acceso remoto",
    "La factura vence el próximo martes",
    "Debo pagar el recibo antes del día 15",
]

modelo_ingles = SentenceTransformer("all-MiniLM-L6-v2")
v = modelo_ingles.encode(frases)
s = util.cos_sim(v, v)

for i in range(len(frases)):
    for j in range(i+1, len(frases)):
        print(f"{s[i][j]:.2f} | {frases[i]}  VS  {frases[j]}")
```

## Experimento 2 — modelo multilingüe


```python
modelo = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
v = modelo.encode(frases)
s = util.cos_sim(v, v)
for i in range(len(frases)):
    for j in range(i+1, len(frases)):
        print(f"{s[i][j]:.2f} | {frases[i]}  VS  {frases[j]}")
```

## Similitud coseno calculada a mano


```python
import numpy as np
a = modelo.encode("No puedo conectarme a la VPN")
b = modelo.encode("Tengo problemas con el acceso remoto")
score = np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
print("A mano:", round(float(score),4))
print("Librería:", round(float(util.cos_sim(a,b)),4))
```

## Dilución


```python
frases_dilucion = [
    "VPN",
    "Usar la VPN",
    "El empleado debe usar la VPN para acceder a sistemas internos",
    "La política general de trabajo remoto incluye requisitos de seguridad, dispositivos administrados, autenticación, actualizaciones y el uso de VPN para algunos recursos internos."
]
q = modelo.encode("VPN")
for t in frases_dilucion:
    print(round(float(util.cos_sim(q, modelo.encode(t))),3), "|", t)
```
